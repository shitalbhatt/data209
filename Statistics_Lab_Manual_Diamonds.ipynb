## 4. Mean vs Median Comparison

- **Mean** = sum of values / number of values (average). Gets affected by very large or
  very small values (outliers).
- **Median** = the middle value when data is sorted. Not affected by outliers.

Rule of thumb:
- Mean ≈ Median → data is symmetric.
- Mean > Median → data is right-skewed (some very high values pull the mean up).
- Mean < Median → data is left-skewed (some very low values pull the mean down).

```python
# Mean and Median of price
mean_price = df["price"].mean()
median_price = df["price"].median()

print("Mean of price  :", mean_price)
print("Median of price:", median_price)

if mean_price > median_price:
    print("Interpretation: Mean > Median -> data is RIGHT-SKEWED")
elif mean_price < median_price:
    print("Interpretation: Mean < Median -> data is LEFT-SKEWED")
else:
    print("Interpretation: Mean = Median -> data is SYMMETRIC")
```

```python
# Let's also check Mean vs Median for carat
mean_carat = df["carat"].mean()
median_carat = df["carat"].median()

print("Mean of carat  :", mean_carat)
print("Median of carat:", median_carat)
```

## 5. Variance vs IQR (Interquartile Range)

- **Variance** and **Standard Deviation** measure how spread out the data is from the
  mean. They use every value, so outliers affect them a lot.
- **IQR = Q3 - Q1** measures the spread of the middle 50% of the data. It is not
  affected much by outliers.

A big difference between what standard deviation suggests and what IQR suggests usually
means the data has outliers.

```python
# Variance and Standard Deviation
variance_price = df["price"].var()
std_price = df["price"].std()

print("Variance of price          :", variance_price)
print("Standard Deviation of price:", std_price)
```

```python
# IQR calculation
Q1 = df["price"].quantile(0.25)
Q3 = df["price"].quantile(0.75)
IQR = Q3 - Q1

print("Q1 (25th percentile):", Q1)
print("Q3 (75th percentile):", Q3)
print("IQR (Q3 - Q1)       :", IQR)
```

```python
# Finding outliers using the 1.5 x IQR rule
lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outliers = df[(df["price"] < lower_limit) | (df["price"] > upper_limit)]

print("Lower limit:", lower_limit)
print("Upper limit:", upper_limit)
print("Number of outliers found:", len(outliers))
```

## 6. Skewness and Distribution Shape

**Skewness** tells us if the data is symmetric or leans to one side.

| Skewness value | Meaning |
|---|---|
| close to 0 | Symmetric distribution |
| greater than 0 | Right-skewed (long tail on the right) |
| less than 0 | Left-skewed (long tail on the left) |

```python
# Skewness of price
skewness_price = df["price"].skew()
print("Skewness of price:", skewness_price)

if skewness_price > 0:
    print("Interpretation: Right-skewed distribution (long tail towards high prices)")
elif skewness_price < 0:
    print("Interpretation: Left-skewed distribution (long tail towards low prices)")
else:
    print("Interpretation: Symmetric distribution")
```

```python
# Skewness of a few more columns
print("Skewness of carat:", df["carat"].skew())
print("Skewness of depth:", df["depth"].skew())
print("Skewness of table:", df["table"].skew())
```

## 7. Histogram, Boxplot, and Density Plot

- **Histogram** shows how frequently values occur, using bars.
- **Boxplot** shows the median, quartiles, and outliers.
- **Density plot** is a smooth curve version of the histogram.

```python
# Histogram of price
plt.figure(figsize=(8, 5))
plt.hist(df["price"], bins=50, color="skyblue", edgecolor="black")
plt.title("Histogram of Diamond Price")
plt.xlabel("Price ($)")
plt.ylabel("Frequency")
plt.show()
```

```python
# Boxplot of price
plt.figure(figsize=(6, 5))
sns.boxplot(y=df["price"], color="lightgreen")
plt.title("Boxplot of Diamond Price")
plt.ylabel("Price ($)")
plt.show()
```

```python
# Density plot of price
plt.figure(figsize=(8, 5))
sns.kdeplot(df["price"], fill=True, color="orange")
plt.title("Density Plot of Diamond Price")
plt.xlabel("Price ($)")
plt.show()
```

**Interpretation:** The histogram and density plot both show that most diamonds are
low to medium priced, with a long tail of very expensive diamonds on the right side —
this matches the positive skewness value we calculated earlier. The boxplot shows many
points above the upper whisker, confirming the presence of outliers (very expensive
diamonds).

## 8. Pivot Tables

A pivot table helps us summarize data by groups. Here we will build pivot tables to see
how average price changes with `cut` and `clarity`, which are categorical columns
already present in the dataset.

```python
# Step 1: Check the categories available in 'cut' and 'clarity'
print("Cut categories    :", df["cut"].unique())
print("Clarity categories:", df["clarity"].unique())
```

```python
# Step 2: Pivot table - Average price by Cut
pivot1 = pd.pivot_table(df, values="price", index="cut", aggfunc="mean")
pivot1
```

```python
# Step 3: Pivot table - Average price by Cut and Clarity
pivot2 = pd.pivot_table(df, values="price", index="cut", columns="clarity", aggfunc="mean")
pivot2
```

```python
# Step 4: Visualize the pivot table as a heatmap
plt.figure(figsize=(10, 6))
sns.heatmap(pivot2, annot=True, fmt=".0f", cmap="YlOrRd")
plt.title("Average Price by Cut and Clarity")
plt.show()
```

**Interpretation:** Interestingly, the pivot table shows that "Ideal" cut diamonds do
not always have the highest average price. This is because `carat` (weight) has a much
bigger effect on price than `cut` alone, and higher-carat diamonds are not evenly spread
across all cut categories in this dataset.
